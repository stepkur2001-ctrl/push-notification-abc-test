```python
import pandas as pd
from scipy import stats
from statsmodels.stats.multicomp import pairwise_tukeyhsd

df = pd.read_csv('user_metrics.csv', sep=';')

group_a = df[df['group'] == 'A']['opened']
group_b = df[df['group'] == 'B']['opened']
group_c = df[df['group'] == 'C']['opened']

f_stat, p_value = stats.f_oneway(group_a, group_b, group_c)
tukey = pairwise_tukeyhsd(endog=df['opened'], groups=df['group'], alpha=0.05)
print(f'ANOVA: F={f_stat:.4f}, p-value={p_value:.6f}')
print(tukey)

print(f'Open rate A: {group_a.mean():.4f}')
print(f'Open rate B: {group_b.mean():.4f}')
print(f'Open rate C: {group_c.mean():.4f}')
```
